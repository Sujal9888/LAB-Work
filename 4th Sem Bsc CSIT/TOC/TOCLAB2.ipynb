```python
# NFA that accepts all binary strings that end with the substring “01” i.e. 
#L = { w | w is a binary string over {0, 1} such that w ends with ‘01’ }. 
# Format: (current_state, input_symbol) : list of possible next states
transitions = {
    ('q0', '0'): ['q0', 'q1'], # stay in q0, AND guess this 0 starts "01"
    ('q0', '1'): ['q0'], # just keep waiting
    ('q1', '1'): ['q2'], # the guessed 0 is followed by 1 -> accept
    # q2 has no outgoing transitions: if more symbols come after q2,
    # that particular guess simply dies out (other guesses may still be alive)
}
start_state = 'q0'
accept_states = ['q2']
# Take the binary string from the user
input_string = input("Enter a binary string (0s and 1s only): ")
# An NFA can be in several states at once, so we track a SET of active states
current_states = {start_state}
print("Start states:", current_states)
# Read the string one character at a time
for symbol in input_string:
    next_states = set()
    for state in current_states:
        if (state, symbol) in transitions:
            next_states.update(transitions[(state, symbol)])
    current_states = next_states
    
    print("Read", symbol, "-> active states:", current_states)
print("Final states:", current_states)
# The NFA accepts if AT LEAST ONE active state is an accept state
if current_states & set(accept_states):
    print("Result: ACCEPTED")
else:
    print("Result: REJECTED")
```

```python
#Construct a NFA for language (ab*a ∪ b*aa).
# Format: (current_state, input_symbol): list of possible next states

transitions = {
    ('q0', 'a'): ['q1'],
    ('q0', ''): ['q3'],      # ε-transition from q0 to q3
    ('q1', 'a'): ['q2'],
    ('q1', 'b'): ['q1'],
    ('q3', 'a'): ['q4'],
    ('q3', 'b'): ['q3'],
    ('q4', 'a'): ['q2']
}

start_state = 'q0'
accept_states = {'q2'}

# Function to compute ε-closure
def epsilon_closure(states):
    closure = set(states)
    stack = list(states)

    while stack:
        state = stack.pop()

        if (state, '') in transitions:
            for next_state in transitions[(state, '')]:
                if next_state not in closure:
                    closure.add(next_state)
                    stack.append(next_state)

    return closure

# Take input from the user
input_string = input("Enter a string over {a,b}: ")

# Start with ε-closure of the start state
current_states = epsilon_closure({start_state})

print("Start states:", current_states)

# Process each symbol
for symbol in input_string:
    next_states = set()

    for state in current_states:
        if (state, symbol) in transitions:
            next_states.update(transitions[(state, symbol)])

    # Compute ε-closure after each move
    current_states = epsilon_closure(next_states)

    print(f"Read '{symbol}' -> Active states: {current_states}")

print("Final states:", current_states)

# Check acceptance
if current_states & accept_states:
    print("Result: ACCEPTED")
else:
    print("Result: REJECTED")
```
