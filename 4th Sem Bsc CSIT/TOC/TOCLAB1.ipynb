```python
# Language accepted: binary strings that end with '01'.
# Format: (current_state, input_symbol) : next_state
transitions = {
('q0', '0'): 'q1',
('q0', '1'): 'q0',
('q1', '0'): 'q1',
('q1', '1'): 'q2',
('q2', '0'): 'q1',
('q2', '1'): 'q0',
}
start_state = 'q0' # where the DFA begins
accept_states = ['q2'] # the DFA accepts if it ends in one of these
# Take the binary string from the user
input_string = input("Enter a binary string (0s and 1s only): ")
# Begin at the start state
current_state = start_state
print("Start state:", current_state)
# Read the string one character at a time
for symbol in input_string:
    current_state = transitions[(current_state, symbol)]
    print("Read", symbol, "-> moved to state", current_state)
print("Final state:", current_state)
# Decide accept or reject
if current_state in accept_states:

    print("Result: ACCEPTED")
else:
    print("Result: REJECTED")
```

```python
# DFA: String must contain BOTH 01 and 10 as substrings

transitions = {
    ('q0', '0'): 'q1',
    ('q0', '1'): 'q2',

    ('q1', '0'): 'q1',
    ('q1', '1'): 'q3',   # found 01

    ('q2', '0'): 'q4',   # found 10
    ('q2', '1'): 'q2',

    ('q3', '0'): 'q5',   # now both 01 and 10 found
    ('q3', '1'): 'q3',

    ('q4', '0'): 'q4',
    ('q4', '1'): 'q5',   # now both 10 and 01 found

    ('q5', '0'): 'q5',
    ('q5', '1'): 'q5'    # accepting state forever
}

# Start state
start_state = 'q0'

# Accepting state
accept_states = ['q5']

# Take input from user
input_string = input("Enter a binary string (0s and 1s only): ")

# Begin at start state
current_state = start_state
print("Start state:", current_state)

# Read the string one character at a time
for symbol in input_string:
    current_state = transitions[(current_state, symbol)]
    print("Read", symbol, "-> moved to state", current_state)

print("Final state:", current_state)

# Decide accept or reject
if current_state in accept_states:
    print("Result: ACCEPTED")
else:
    print("Result: REJECTED")
```
