```python
# OR Gate using a Single Neuron
def step(x):
    if x >= 0:
        return 1
    else:
        return 0
# Inputs and expected outputs
inputs = [
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
]

expected = [0, 1, 1, 1]
# Weights and bias
w1 = 1
w2 = 1
bias = -0.5
print("Sujal Meher Kayastha\n")
print("X1 X2 | Output")

for x, target in zip(inputs, expected):
     # Weighted sum
    net = x[0] * w1 + x[1] * w2 + bias

    # Activation function
    output = step(net)

    print(x[0], " ", x[1], " | ", output)
```
