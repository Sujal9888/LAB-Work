```python
# AND Gate using Adaline

def threshold(z):
    if z >= 0:
        return 1
    else:
        return -1


X = [[-1, -1], [-1, 1], [1, -1], [1, 1]]
Y = [-1, -1, -1, 1]

learning_rate = 0.1

# Initial weights
w1 = 0
w2 = 0
b = 1

# Adaline Learning
for epoch in range(100):
    for i in range(4):

        # Calculate output
        z = X[i][0] * w1 + X[i][1] * w2 + b

        # Calculate error
        error = Y[i] - z

        # Update weights and bias
        w1 = w1 + learning_rate * error * X[i][0]
        w2 = w2 + learning_rate * error * X[i][1]
        b = b + learning_rate * error


print("Sujal Meher Kayastha")
print("Final weights:", w1, w2)
print("Final bias:", b)

print("\n X1   X2    Y")

for x1 in [-1, 1]:
    for x2 in [-1, 1]:
        y = threshold(x1 * w1 + x2 * w2 + b)
        print(f"{x1:>3}  {x2:>3}  {y:>3}")
```
