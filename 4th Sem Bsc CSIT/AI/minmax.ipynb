```python
# Minimax Algorithm

def minimax(tree, maximizing):
    if not isinstance(tree, list):
        return tree

    values = []
    
    for child in tree:
        values.append(minimax(child, not maximizing))

    if maximizing:
        return max(values)
    else:
        return min(values)


# Game Tree
tree = [
    [[3, 5], [2, 9]],
    [[12, 5], [23, 7]]
]

result = minimax(tree, True)
print("Sujal Meher Kayastha")
print("Best value:", result)
```
