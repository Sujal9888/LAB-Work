```python
import heapq

graph = {
    'A': [('B', 1), ('C', 1)],
    'B': [('D', 1), ('E', 1)],
    'C': [('F', 1), ('G', 1)],
    'D': [],
    'E': [('F', 1)],
    'F': [],
    'G': []
}

# Heuristic values (estimated distance to goal G)
heuristic = {
    'A': 2,
    'B': 2,
    'C': 1,
    'D': 3,
    'E': 2,
    'F': 1,
    'G': 0
}

def astar(graph, start, goal):
    pq = []
    heapq.heappush(pq, (heuristic[start], 0, start))  # (f, g, node)

    visited = set()

    while pq:
        f, g, node = heapq.heappop(pq)

        if node in visited:
            continue

        print(node, end=" ")
        visited.add(node)

        if node == goal:
            print("\n-> Goal Node Found!")
            return

        for neighbour, cost in graph[node]:
            if neighbour not in visited:
                new_g = g + cost
                new_f = new_g + heuristic[neighbour]
                heapq.heappush(pq, (new_f, new_g, neighbour))

# Driver code
print("Sujal Meher Kayastha")
astar(graph, 'A', 'G')
```
