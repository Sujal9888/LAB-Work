```python
# AI: Simple Reflex Agent and Goal-Based Agent

# ---------------- Simple Reflex Agent ----------------
def simple_reflex_agent(location, status):
    if status == "Dirty":
        return "Suck"
    elif location == "A":
        return "Move Right"
    else:
        return "Move Left"


# ---------------- Goal-Based Agent ----------------
def goal_based_agent(location, status, goal):
    if status == "Dirty":
        return "Suck"

    if goal == "Clean":
        if location == "A":
            return "Move Right"
        elif location == "B":
            return "Move Left"

    return "No Action"

print("\n---Sujal Meher Kayastha---")

# ---------------- Main Program ----------------
print("----- SIMPLE REFLEX AGENT -----")

location = input("Enter location (A/B): ")
status = input("Enter status (Clean/Dirty): ")

action = simple_reflex_agent(location, status)

print("Location:", location)
print("Status:", status)
print("Action:", action)


print("\n----- GOAL-BASED AGENT -----")

location = input("Enter location (A/B): ")
status = input("Enter status (Clean/Dirty): ")
goal = input("Enter goal (Clean): ")
action = goal_based_agent(location, status, goal)

print("Location:", location)
print("Status:", status)
print("Goal:", goal)
print("Action:", action)
```
