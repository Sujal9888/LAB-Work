```python
def threshold(z):
    if z >= 0.5:
        return 1
    else:
        return 0


def OR(x1, x2, w1, w2):
    z = x1 * w1 + x2 * w2
    return threshold(z)


w1 = 0.6
w2 = 0.6
print("Sujal Meher Kayastha")
print("X1  X2  Y")

for x1 in range(2):
    for x2 in range(2):
        y = OR(x1, x2, w1, w2)
        print(x1, " ", x2, " ", y)
```
