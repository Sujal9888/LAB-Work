```python
# DFS to find Goal Node and Path

graph = {
    'S': ['A', 'B', 'C'],
    'A': ['D', 'E'],
    'B': ['G'],
    'C': ['F'],
    'D': ['H'],
    'E': ['G'],
    'F': ['G'],
    'G': [],
    'H': []
}


def DFS(node, goal, visited, path):
    visited.add(node)
    path.append(node)

    if node == goal:
        return True

    for child in graph[node]:
        if child not in visited:
            if DFS(child, goal, visited, path):
                return True

    path.pop()
    return False


visited = set()
path = []

DFS('S', 'G', visited, path)

print("Sujal Meher Kayastha")
print("Goal Node:", 'G')
print("DFS Path:", " -> ".join(path))

print("\nDFS Performance:")
print("Time Complexity: O(b^m)")
print("Space Complexity: O(bm)")
```

```python
# OR Gate using Hebbian Learning

def threshold(z):
    if z >= 0:
        return 1
    else:
        return -1


def OR(x1, x2, w1, w2, b):
    z = x1 * w1 + x2 * w2 + b
    return threshold(z)


X = [[-1, -1], [-1, 1], [1, -1], [1, 1]]
Y = [-1, 1, 1, 1]

# Initial weights and bias
w1 = 0
w2 = 0
b = 1

# Hebbian learning
for i in range(4):
    w1 = w1 + X[i][0] * Y[i]
    w2 = w2 + X[i][1] * Y[i]
    b = b + Y[i]

print("Sujal Meher Kayastha")
print("Final weights:", w1, w2)
print("Final bias:", b)

print("\n X1   X2    Y")

for x1 in [-1, 1]:
    for x2 in [-1, 1]:
        y = OR(x1, x2, w1, w2, b)
        print(f"{x1:>3}  {x2:>3}  {y:>3}")
```
