```python
# XOR Gate using Backpropagation

import math

def sigmoid(x):
    return 1 / (1 + math.exp(-x))


def sigmoid_derivative(x):
    return x * (1 - x)


# Input and target
X = [[0, 0], [0, 1], [1, 0], [1, 1]]
Y = [0, 1, 1, 0]

# Initial weights and biases
w1, w2 = 0.1, 0.2
w3, w4 = 0.3, 0.4
w5, w6 = 0.5, 0.6

b1, b2, b3 = 0.1, 0.1, 0.1

learning_rate = 0.5

# Training
for epoch in range(10000):

    for i in range(4):

        x1, x2 = X[i]
        target = Y[i]

        # Forward propagation
        h1 = sigmoid(x1 * w1 + x2 * w2 + b1)
        h2 = sigmoid(x1 * w3 + x2 * w4 + b2)

        output = sigmoid(h1 * w5 + h2 * w6 + b3)

        # Error
        error = target - output

        # Backpropagation
        delta_output = error * sigmoid_derivative(output)

        delta_h1 = delta_output * w5 * sigmoid_derivative(h1)
        delta_h2 = delta_output * w6 * sigmoid_derivative(h2)

        # Update output weights
        w5 += learning_rate * delta_output * h1
        w6 += learning_rate * delta_output * h2
        b3 += learning_rate * delta_output

        # Update hidden weights
        w1 += learning_rate * delta_h1 * x1
        w2 += learning_rate * delta_h1 * x2
        b1 += learning_rate * delta_h1

        w3 += learning_rate * delta_h2 * x1
        w4 += learning_rate * delta_h2 * x2
        b2 += learning_rate * delta_h2


# Testing
print("Sujal Meher Kayastha")
print("\nX1  X2  Output")

for x1, x2 in X:

    h1 = sigmoid(x1 * w1 + x2 * w2 + b1)
    h2 = sigmoid(x1 * w3 + x2 * w4 + b2)

    output = sigmoid(h1 * w5 + h2 * w6 + b3)

    if output >= 0.5:
        result = 1
    else:
        result = 0

    print(x1, "  ", x2, "  ", result)
```
