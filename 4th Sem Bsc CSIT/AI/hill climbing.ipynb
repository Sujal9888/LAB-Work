```python
# Hill Climbing Algorithm
a = [
    [5, 9, 1],
    [7, 2, 3],
    [8, 6, 4]
]

current = 5

while True:
    # Find all neighboring values
    neighbors = []

    for i in range(3):
        for j in range(3):
            if a[i][j] == current:

                if i > 0:
                    neighbors.append(a[i-1][j])

                if i < 2:
                    neighbors.append(a[i+1][j])

                if j > 0:
                    neighbors.append(a[i][j-1])

                if j < 2:
                    neighbors.append(a[i][j+1])
     # Find the best neighbor
    best = max(neighbors)

    if best > current:
        current = best
    else:
        break
print("Sujal Meher Kayastha")
print("Local Maximum:", current)
```
